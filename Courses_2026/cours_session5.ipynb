## 1) Metrics 

### Accuracy :

What is accuracy ? It is the amount of correct answers. On the confusion matrix, it represents the percentage of True prediction from the total amount of prediction attempts.

**Formula :** Accuracy = (TP + TN) / (TP + TN + FP + FN) = Correct / 
Total

When to use ✓
- Balanced classes (50/50 split)
- Quick overall performance check
- When all errors are equally bad

When NOT to use ✗
- Imbalanced data 
- When some errors are worse than others
- Medical diagnosis, fraud detection, rare events
 

### Precision :

Precision is from the total of true predictions, the percentage that was actually true. 

**Formula :** Precision = TP / (TP + FP)

Also called: Positive Predictive Value (PPV)

**Example (Cancer):** Precision = 90 / (90 + 50) 
= 90/140 = 64.3%
Interpretation: "When we say 'cancer', we're 
right 64% of the time"

#### **PRIORITIZE PRECISION WHEN:**
False Positives are COSTLY!
Spam Detection
   FP = Important email marked as spam
   (You miss a job offer!)
Criminal Justice
   FP = Innocent person convicted
   (Terrible injustice!)
Marketing Targeting
   FP = Wasted ads on uninterested people
   (Money down the drain!)
